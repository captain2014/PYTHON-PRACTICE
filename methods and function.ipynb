```python
def check_even_list(num_list):
    for num in num_list:
        if num%2==0:
            return True
        else:
            pass
```

```python
check_even_list([1,1,2])
```

```python
def check_even_list(num_list):
    even_num=[]
    
    for num in num_list:
        if num%2==0:
            even_num.append(num)
        else:
            pass
    return even_num
```

```python
check_even_list([3,5,9])
```

```python
check_even_list([1,2,3,4,5,6,7,8,9,])
```

```python
wrk_hrs=[('a',5),('b',6),('c',2),('d',1)]
```

```python
def emp_check(wrk_hrs):
    current_max=0
    emp_of_month=''
    for x,y in wrk_hrs:
        if y>current_max:
            current_max=y
            emp_of_month=x
        else:
            pass
    return(emp_of_month,current_max)
```

```python
emp_check(wrk_hrs)
```

```python
result=emp_check(wrk_hrs)
```

```python
result
```

```python
x,y=emp_check(wrk_hrs)
```

```python
def player_guess():
    guess=''

    while guess not in ['0','1','2']:
        guess=input('pick num 0,1 0r 2')
    return int(guess)
```

```python
player_guess()
```

```python
myindex=player_guess()
```

```python
myindex
```

```python
def check_guess(mylist,guess):
    if mylist[guess]=='0':
        print('correct')
    else:
        print('wrong!')
    
```

```python
#initial list
mylist=['','o','']
#shuffle list
mixedup_list=shuffle_list(mylist)

#user guess
guess=player_guess()
#check guess
check_guess(mixedup_list,guess)
```

```python
#using *args

def func(*args):
    return sum(args)*0.05
```

```python
func(1,2,3,4,5,6,7,8,9)
```

```python
#using **kwargs

def func(**kwargs):
    if 'fruit' in kwargs:
        print('my fruit choice is {}'.format(kwargs['fruit']))
    else:
        print('no choice')

```

```python
func(fruit='apple')
```

```python
def func(*args,**kwargs):
    print(args)
    print(kwargs)
    print('i would like {} {} '.format(args[0],kwargs['food']))
```

```python
func(10,20,30,food='her', fruit='her')
```
